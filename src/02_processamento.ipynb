```python
from pyspark.sql import SparkSession

# 1. Inicializando o motor do Big Data (PySpark)
spark = SparkSession.builder \
    .appName("TCC_Bacen_Processamento") \
    .getOrCreate()

print("✅ Spark iniciado com sucesso!\n")

# 2. Lendo todos os CSVs
caminho_raw = "../data/raw/*.csv"
print("⏳ Lendo os arquivos CSV. Isso pode levar alguns segundos...")

# Assumindo ponto-e-vírgula como separador padrão do BACEN
df_scr = spark.read.csv(
    caminho_raw, 
    header=True, 
    sep=";",          
    inferSchema=True, 
    encoding="utf-8"  
)

# 3. Contando as linhas e exibindo a estrutura
print(f"📊 Leitura concluída! Total de linhas carregadas: {df_scr.count()}\n")

print("Estrutura das colunas (Schema):")
df_scr.printSchema()

print("\nPrimeiras 5 linhas dos dados:")
df_scr.show(5, truncate=False)
```
