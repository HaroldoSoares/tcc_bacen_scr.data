```python
import sys
import os
import glob
import pyspark
from pyspark.sql import SparkSession

print("=== 1. VALIDAÇÃO DE INFRAESTRUTURA (HEALTH CHECK) ===")
print(f"Versão do Python: {sys.version.split(' ')[0]}")
print(f"Versão do PySpark: {pyspark.__version__}")

try:
    java_version = os.popen('java -version 2>&1 | awk -F[\\\"_] \'NR==1{print $2}\'').read().strip()
    print(f"Versão do Java Ativa: {java_version}")
except:
    print("Versão do Java: Não foi possível identificar.")

print("\n=== 2. INICIALIZAÇÃO DO MOTOR SPARK ===")
try:
    # Injeção de 6GB de RAM para suportar a carga de 2020-2024
    spark = SparkSession.builder \
        .appName("TCC_HealthCheck_Volumetria") \
        .config("spark.driver.memory", "6g") \
        .getOrCreate()
    print("✅ SparkSession: ATIVA E OPERACIONAL (6GB RAM alocados)")
except Exception as e:
    print(f"❌ ERRO CRÍTICO no Spark: {e}")

print("\n=== 3. MAPEAMENTO E VOLUMETRIA DA CAMADA RAW (2020-2026) ===")
caminho_raw = "../data/raw/"
arquivos_csv = glob.glob(f"{caminho_raw}/*.csv")
print(f"📁 Arquivos físicos encontrados no diretório: {len(arquivos_csv)}")

if len(arquivos_csv) > 0:
    print("⏳ Lendo arquivos e calculando a volumetria total (isso pode levar alguns minutos)...")
    df_raw = spark.read.csv(f"{caminho_raw}/*.csv", sep=";", header=True, encoding="utf-8")
    
    total_linhas = df_raw.count()
    total_colunas = len(df_raw.columns)
    
    print(f"📊 Volumetria Total Consolidada: {total_linhas:,} linhas e {total_colunas} colunas.")
else:
    print("❌ Nenhum arquivo encontrado. Verifique se foram descompactados corretamente.")
```
