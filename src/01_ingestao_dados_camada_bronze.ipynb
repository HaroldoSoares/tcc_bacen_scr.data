```python
import requests
import zipfile
import io
import os

def ingerir_dados_bacen(ano, pasta_destino="../data/raw/"):
    """
    Faz o download do arquivo ZIP do BACEN referente ao ano especificado
    e extrai os CSVs diretamente para a pasta de destino.
    """
    url = f"https://www.bcb.gov.br/pda/desig/scrdata_{ano}.zip"
    
    # Garante que a pasta destino existe na raiz do projeto, se não, ele cria
    os.makedirs(pasta_destino, exist_ok=True)
    
    print(f"[{ano}] Iniciando o download da URL: {url}")
    print(f"[{ano}] Aguarde, baixando o arquivo compactado...")
    
    # Fazendo a requisição GET
    resposta = requests.get(url, stream=True)
    
    # Verifica se a requisição foi um sucesso (Status 200)
    if resposta.status_code == 200:
        print(f"[{ano}] Download concluído! Extraindo arquivos CSV...")
        
        # Descompacta o arquivo ZIP em memória e salva na pasta raw
        with zipfile.ZipFile(io.BytesIO(resposta.content)) as zip_ref:
            zip_ref.extractall(pasta_destino)
            
        print(f"[{ano}] Sucesso! Arquivos extraídos em: {pasta_destino}")
    else:
        print(f"[{ano}] Erro ao baixar os dados. Status Code: {resposta.status_code}")

# ==========================================
# EXECUÇÃO EM LOTE (BATCH INGESTION)
# ==========================================

# Definindo o escopo histórico desde 2020 até 2026
anos_historicos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

print("Iniciando a esteira de ingestão de dados em lote...\n")

for ano in anos_historicos:
    print("-" * 50)
    ingerir_dados_bacen(ano)

print("\n🚀 Ingestão da série histórica completa concluída com sucesso!")
```
